## Objective

The objective of this analysis is to understand customer characteristics,
customer types, ordering behavior, and feedback using Python and Pandas.

## Dataset

The dataset contains 285 customer records and 13 attributes.

## Analysis Performed

1. Analyzed customer age distribution.
2. Analyzed customers by gender.
3. Analyzed customer types.
4. Analyzed occupations.
5. Analyzed monthly income groups.
6. Analyzed customer feedback.
7. Compared customer type with output.
8. Compared gender with output.
9. Compared occupation with output.

## Key Findings

- The dataset contains 285 customer records.
- Customer age distribution was analyzed using descriptive statistics.
- Gender distribution was analyzed.
- Regular and frequent customers were compared.
- Customer feedback was analyzed to understand satisfaction.
- Customer type and output were compared using a cross-tabulation.
- Gender and occupation were also compared with customer output.
